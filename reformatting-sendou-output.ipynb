```python
print("Hell World")

input_file_path = "Testing/all-participants (1).txt"
write_file_path = "Testing/sendou_scrap.csv"
```

```python
START_TEAM_TOKEN = "-- "
END_TEAM_TOKEN = " --"

CAPTAIN_TOKEN = "(C) "

START_PLAYER_TOKEN = "(IGN: "
END_PLAYER_TOKEN = ") - "
```

```python
TEAM_NAME_BATTLEFY_HEADER = 'teamName'
PLAYER_NAME_BATTLEFY_HEADER = 'inGameName'
PLAYER_DISCORD_BATTLEFY_HEADER  = 'Captain\'s Discord username'
```

```python
with open("Testing/sendou_scrap.csv", "w") as write_file:
    first_line = TEAM_NAME_BATTLEFY_HEADER + "," + PLAYER_NAME_BATTLEFY_HEADER + "," + PLAYER_DISCORD_BATTLEFY_HEADER + "\n"

    write_file.write(first_line)


    with open(input_file_path) as file:
        current_team, current_captain = "", ""
        
        for line in file:
            line = line.replace("\n", "")
            line = line.replace("\"", "\"\"")

            if line.strip():

                if START_TEAM_TOKEN in line and END_TEAM_TOKEN:
                    # New Team
                    current_team = line.replace(START_TEAM_TOKEN, "").replace(END_TEAM_TOKEN, "")

                else:
                    # Player
                    if CAPTAIN_TOKEN in line:
                        line = line.replace(CAPTAIN_TOKEN, "")
                        current_captain = line[line.rfind(END_PLAYER_TOKEN) + len(END_PLAYER_TOKEN):]
                    
                    player = line[line.rfind(START_PLAYER_TOKEN) + len(START_PLAYER_TOKEN): line.rfind(END_PLAYER_TOKEN)]

                    write_file.write(current_team + ",\"" + player  + "\"," + current_captain + "\n")
                    print(current_team, player, current_captain, sep=",")
```
